### Problem 3.1: Data Processing
Your team has collected some survey data. Help clean and process it:


```python
# Raw survey responses (some have extra spaces, inconsistent capitalization)
responses = ["yes", "NO", "maybe", "YES", "no", "Maybe", "yes"]

# Write code to:
# 1. Clean each response (convert to lowercase)
# 2. Count how many "yes", "no", and "maybe" responses there are
# 3. Create separate lists for each type of response
# 4. Create separate tuples for each type of response

lowerNames = []
for response in responses:
    lowerNames.append(response.lower())

yesCount = responses.count("yes")
noCount = responses.count("no")
maybeCount = responses.count("maybe")

print(responses)
print(yesCount)
print(noCount)
print(maybeCount)

yesList = [lowerNames[0], lowerNames[3],lowerNames[6]]
print(yesList)

noList = [lowerNames[1], lowerNames[4]]
print(noList)

maybeList = [lowerNames[2], lowerNames[5]]
print(maybeList)

yesTuple = (lowerNames[0], lowerNames[3],lowerNames[6])
print(yesTuple)

noTuple = (lowerNames[1], lowerNames[4])
print(noTuple)

maybeTuple = (lowerNames[2], lowerNames[5])
print(maybeTuple)
```

### Problem 3.2: Text Analysis


```python
# Analyze this text:
text = """Monty Python's sketch on spam https://en.wikipedia.org/wiki/Spam_(Monty_Python_sketch)
directly led to the use of spam to describe unwanted email. As a joke, and in reference
to therepetitive and unwanted presence of spam in the sketch, early internet users
flooded forums with the word SPAM . This came to describe the equally unwanted marketing
and repetetive emails which flood our inbox daily."""



# Write code to:
# 1. Split the text into individual words
# 2. Count how many words there are total
# 3. Count how many times "spam" appears, both upper and lowercase

textSplit = text.split()
print(textSplit)
print(len(textSplit))

lowerSpams = []
for response in textSplit:
    lowerSpams.append(response.lower())

print(lowerSpams.count("spam"))
print(lowerSpams)
```

## Part 4: Challenge Problem
This is a bit of a stretch. You will have to do some work on your own to find the additional command necessary to test if something is a string or a number.


```python
# You have a list of mixed data types:
mixed_data = [1, "hello", 2.5, "world", 42, "python", 3.14]

# Write code to:
# 1. Create two separate lists: one for numbers, one for strings
# 2. Calculate the sum of all numbers
# 3. Join all strings together with spaces between them

numbers = []
strings = []

for item in mixed_data:
    if isinstance(item, (int, float)):
        numbers.append(item)
    elif isinstance(item, str):
        strings.append(item)

print(numbers)
print(strings)
```
