## Problem 2: Formatted output with nested loops

Your task is to practice nested loops by writing a Python program that generates a simple pattern.

- Loop over the numbers 0 through 4.
- On each line, first print the number, followed by a colon (:).
- After the colon, print as many asterisks (*) as the number itself.
- Each line should show the number and its matching stars.

Example Output:

```
0:
1:*
2:**
3:***
4:****
```

Hint: You will need a loop inside a loop — the outer loop handles the numbers, and the inner loop handles printing the stars.

````python
for i in range(5):
    print(i, end = ':')
    for j in range(i):
        print('*', end = '')
    print()
````

## Problem 3: Looping Through a List to Print Element Types

Write a Python program that prints the type of each element in a list.

## Problem 4: Printing Exclamation Point Patterns

Write a Python program that uses *nested* `for` loops to print rows of exclamation points (`!`) based on a list of positive integers.

Requirements:

- For each number in the list, print a row of exclamation points equal to that number.
  - Example: if the number is `3`, print `!!!`.
- Each row should have no spaces between the exclamation points.
- Write your solution in two ways:
  - Approach 1: Use the outer loop to go through the elements of the list directly.
  - Approach 2: Use the outer loop to go through the indices of the list, and access elements by indexing.

Example List:

```python
intlist = [3, 7, 2, 1, 5]
```

````python
intlist = [3, 7, 2, 1, 5]

for i in intlist:
    for j in range(i):
        print('!', end = '')
    print()

print('\n\n')

for i in range(len(intlist)):
    for j in range(intlist[i]):
        print('!', end = '')
    print()
````

## Problem 5: Printing Characters from Strings with Separators

Use the following list of strings. Write code to print all of the characters from all of the strings on one line, with a dash (`-`) after each.

```python
strlist = ['I', 'love', 'coding!']
```

## Problem 6: Nested `while` and `for` loops

Write a script that will prompt the user for positive numbers. As long as the user is entering positive numbers, Print `Ni!` that many times on a single line for each. For example, if the user enters 3, `Ni!Ni!Ni!` would be printed.

````python
posnum = int(input('Enter a positive number: '))

while posnum > 0:
    for i in range(posnum):
        print('Ni!', end = '')
    print()
    posnum = int(input('Enter a positive number: '))
````

## Problem 7: Triangle of stars

Prompt the user for a positive integer `n`. Then print a triangle of stars, starting with 0 in the first row and incrementing to `n-1` stars in the `n-1st` row.

Example output for `n=4`

```
*
**
***
```

````python
n = int(input('Enter a positive integer: '))
while n <= 0:
    n = int(input('Enter a positive integer: '))

for i in range(n):
    for j in range(i):
        print('*', end='')
    print()
````

## Problem 8: Simulating Multiple Trials with Random Numbers

Engineers often need to test systems multiple times with random inputs to find bugs or verify performance.

Write a program that:
- Uses `random.seed(42)` for reproducibility
- Simulates **3 trials**
- In each trial, generate **4 random integers** between -5 and 5 (inclusive)
- Print all 4 numbers on one line for each trial
- Use nested loops (outer loop for trials, inner loop for random numbers)

**Expected Output** (with seed=42):
```
Trial 1: 0 -4 -1 3
Trial 2: 3 -5 3 -4
Trial 3: -5 4 1 -3
```

**Hint:** Import the random module and use `random.randint(-5, 5)` to generate each number.

````python
import random

random.seed(42)  # Set seed for reproducibility

for trial in range(3):
    print(f'Trial {trial+1}:', end=' ')
    for i in range(4):
        num = random.randint(-5, 5)
        print(num, end=' ')
    print()  # Move to next line after each trial
````
