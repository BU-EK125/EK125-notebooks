## 1. Strings — Escape Characters

### • Escape sequences (`\n`, `\t`, `\\`) tell Python "this isn't a backslash, it's an instruction"

### • `print()` shows real spacing; `repr()` re-escapes it back to `\t` — same string, two views

````python
# From the reading's "Escape Characters" section (Class4.html)
print("Hello\nWorld")
print("Column1\tColumn2")
print("A backslash looks like this: \\")
"Column1\tColumn2"
````

## Indexing

- Every character sits at a numbered position — grab one with `word[i]`
- Counting starts at `0`, not `1`, and can run backward from the end
- `word2[-1]` is always the last character — safer than `word2[len(word2) - 1]`

````python
# From the reading's "Indexing" section (Class4.html)
word2 = "Python"
print(word2[0])  # 'P'
print(word2[-1])  # 'n'
word2[-1]
````

## Slicing

- Slicing gets a whole chunk: `word[start:stop]` (leave off either side for "beginning"/"end")
- **Excludes the end index** — `word[0:2]` stops *before* position `2` → `'Py'`, not `'Pyt'`
- The number after the colon is where the slice *stops*, not the last character included

````python
# From the reading's "Slicing" section (Class4.html)
word = "Python"
print(word[0:2])  # 'Py'
print(word[:3])  # 'Pyt'
print(word[3:])  # 'hon'
word[0:2]
````

## REMEMBER: Strings Are Immutable!

- Once a string is created, none of its characters can be changed — `s[0] = "b"` raises an error
- Fix: slice out what you want, glue it back with `+` into a *new* string
- Reassigning `s` doesn't mutate the old string — it points the name at a different one

````python
# From the reading — "Strings Are Immutable" (Class4.html)
s = "cat"
# s[0] = "b"  # ERROR: strings don't support assignment
s = "b" + s[1:]
s
````

## Common Mistakes with Strings

- Classic mistake: trying to assign directly into a character position

```python
s = "dog"
s[0] = "c"   # trying to mutate a string in place
```
```
TypeError: 'str' object does not support item assignment
```

- Recognize this exact error on sight — it's what you'll see in your own terminal

## 2. Objects and Methods

- Strings and lists aren't just raw data — they're **objects**
- Objects come with built-in **methods**: functions that belong to the value, called with a dot

````python
# From the reading's "Methods for Strings" (Class4.html)
word3 = "python"
print(word3.upper())  # 'PYTHON'
print(word3.capitalize())  # 'Python'
print(word3.replace("py", "java"))  # 'javathon'
word3.find("th")
````

- `.upper()` is exactly what GPP Problem 1.1 needs for the report header — `"steel".upper()` → `"STEEL"`
- None of these methods change `word3` itself — each hands back a new string (strings are immutable)

````python
# From the reading's "Methods for Lists" (Class4.html)
numbers_list = [3, 1, 4]
numbers_list.append(2)
numbers_list.sort()
numbers_list.reverse()
numbers_list
````

- `.append()`, `.sort()`, `.reverse()` all mutate `numbers_list` **in place** — no reassignment needed
- Lists are mutable (unlike strings)
- Same list methods from Class 2, just showing up again in review

## Common Errors with Methods

- A method must be *called* with `()` to do anything — leaving them off is an easy typo

```python
word = "python"
print(word.upper)   # forgot the parentheses!
```
```
<built-in method upper of str object at 0x...>
```

- Without `()`, Python hands you a *reference* to the method, not its result

## 4. Type Casting (Converting Between Types)

- `input()` always returns a string — "casting" converts it to a number
- Cast **before** doing math, not after
- `int()` → whole number; `float()` → keeps decimals (matters for beam lengths/current in this GPP)

````python
# From the reading's "Type Casting" section (Class4.html)
numeric_age = float("21.0")
type(numeric_age)
#numeric_age + 1
````

## Common Errors with Casting

- Casting only works if the text actually looks like the target type

```python

```

## 5. Output Formatting

## f-Strings

- An f-string drops a variable's value directly into a string: `f"...{value}..."`
- No more `+` and `str()` chains to stitch a message together
- Exactly what GPP's beam and resistor reports are built from — one f-string per line

````python
# From the reading's "f-Strings" section (Class4.html)
name = "Alice"
age = 21
f"My name is {name} and I am {age} years old."
````

## Field Width and Alignment

- f-strings can control exactly how much space a value takes up, and its alignment
- Format spec after the colon: `:8d` = width 8, right-aligned (default); `<8d` = left; `^8d` = centered
- Exactly the beam-length table format GPP Problem 1.1 asks for — width 10, right-aligned, unit `m`

````python
# From the reading's "Field Width and Alignment" section (Class4.html)
width_num = 1234
print(f"{width_num:8d}")  # right-aligned (default)
print(f"{width_num:<8d}")  # left-aligned
print(f"{width_num:^8d}")  # centered
f"{width_num:^8d}"
````

## Common Errors with Output

- Forgetting the `f` prefix — no error, just silently wrong:

```python
name = "Alice"
print("Hello {name}")   # forgot the f — literally prints {name}
```
```
Hello {name}
```

- Wrong format code — this one *does* error:

```python
pi = 3.14
print(f"{pi:d}")   # 'd' is for integers, not floats
```
```
ValueError: Unknown format code 'd' for object of type 'float'
```

- The two easiest ways to break a GPP report table — check both when a table looks off

## Selection Statements

- Selection statements are how a program starts *deciding* what to do with values, not just computing them

```python
if condition:
    action
rest of code
```

- `condition` must be `True`/`False` — same Booleans from Class 3
- Indentation (4 spaces) defines which lines belong to the `if`

## If Statements

- No "else" — just "run this if true, otherwise do nothing extra"
- If `False`, the indented block never runs — no error, nothing dramatic
- Rest of the code continues exactly as if the `if` block weren't there

````python
# From the reading's "If Statements" section (Class4.html)
num = 33
if num < 50:
    print('It is smaller')
print('And that is it')
num
````

## If-else Statements

- `else` adds a second path — one branch runs on `True`, the other on `False`
- Never both, never neither

````python
# From the reading's "If-else Statements" section (Class4.html)
num2 = -4.0
if num2 < 0:
    print(f'{num2} is a negative number')
else:
    print(f'{num2} is a nonnegative number')
num2
````

- `else` catches everything the `if` didn't — exactly one branch runs, every time
- Never have to write out the opposite condition yourself

````python
# From the reading's "If-else Statements" section (Class4.html)
num3 = 0.0
if num3 < 0:
    print(f'{num3} is a negative number')
elif num3 == 0:
    print('It is a zero')
else:
    print(f'{num3} is a positive number')
num3
````

- `elif` adds more branches without nesting a new `if` inside `else`
- Python checks top to bottom, stops at the first `True`
- Later branches are never evaluated once a match is found

````python
# From the reading's "If-else Statements" section (Class4.html)
num4 = 7
if 5 <= num4 <= 10:
    print('In range')
else:
    print('Not in range')
num4
````

- `5 <= num4 <= 10` is `(5 <= num4) and (num4 <= 10)` written more compactly
- Same chained-comparison trick from Class 3's Booleans, now inside an `if`

## Nested `if` vs. `elif`

- This GPP has you check leap years using **three levels of nested if/else**
- Works, but easy to lose track of which branch you're in
- Next: the same decision, written both ways

````python
# Nested if/else — an if inside an if inside an if
grade_score = 82

if grade_score >= 90:
    letter_grade = "A"
else:
    if grade_score >= 80:
        letter_grade = "B"
    else:
        if grade_score >= 70:
            letter_grade = "C"
        else:
            letter_grade = "F"

letter_grade
````

- Every branch after the first nests one level deeper inside the previous `else`
- By the "F" case, you're three indents deep for one decision

````python
# The exact same decision, flattened with elif
if grade_score >= 90:
    letter_grade2 = "A"
elif grade_score >= 80:
    letter_grade2 = "B"
elif grade_score >= 70:
    letter_grade2 = "C"
else:
    letter_grade2 = "F"

letter_grade2
````

- Same `grade_score`, same result (`"B"`) either way
- `elif` keeps every branch at the same indentation level
- Same transformation the reading applies to the leap-year check
